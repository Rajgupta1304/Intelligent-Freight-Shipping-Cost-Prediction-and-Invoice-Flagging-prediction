```python
x = vendor_dataframe[['Dollars']]
```

```python
y = vendor_dataframe[['Freight']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
model1 = LinearRegression()
model1.fit(x_train, y_train)

model2 = DecisionTreeRegressor(random_state = 42)
model2.fit(x_train, y_train)

model3 = RandomForestRegressor(random_state = 42)
model3.fit(x_train, y_train)
```

```python
def evalute_model(model,X_test,Y_test,model_name):
    prediction = model.predict(X_test)

    mae = mean_absolute_error(prediction,Y_test)
    rmse = mean_squared_error(prediction,Y_test)
    r2 = r2_score(prediction,Y_test)*100

    print(f"{model_name} performances:")
    print(f"MAE : {mae:.2f}")
    print(f"RMSE  :{rmse:.2f}")
    print(f"R2  :{r2:.2f}%")
```

```python
evalute_model(model1,x_test,y_test,'linear regression')
evalute_model(model2,x_test,y_test,'decision_tree_regressor')
evalute_model(model3,x_test,y_test,'random-forest-regressor')
```
